# Create stoage volumes
Setting up volumes to store our sample data and streaming files

```python
# create volumns for our sample data
spark.sql("CREATE VOLUME IF NOT EXISTS dev.bronze.sample_data")
print("volumn created dev.bronze.sample_data")
```

```python
# creating volumns for streming input (where simulater will write files)
spark.sql("CREATE VOLUME IF NOt EXISTS dev.bronze.streaming_input")
print("volumn created dev.bronze.streaming_input")
```

```python
# create volumn for checkpoint (streamign job state)
spark.sql("CREATE VOLUME IF NOT EXISTS dev.bronze.checkpoints")
print("volumn created dev.bronze.checkpoints")
```

```python
#List all volumns in bronze schema
display(spark.sql("SHOW VOLUMES IN dev.bronze"))
```
