# Unity Catalog Setup
# Creating catalog and schemas for our IoT project

```python
# Create development catalog
spark.sql("CREATE CATALOG IF NOT EXISTS dev")
print("✅ Catalog 'dev' created")
```

```python
# create bronze schema
spark.sql("CREATE SCHEMA IF NOT EXISTS dev.bronze")
print("✅ Schema 'dev.bronze' created")

# create silver schema
spark.sql("CREATE SCHEMA IF NOT EXISTS dev.silver")
print("Schema 'dev.silver' created")

# create gold schema
spark.sql("CREATE SCHEMA IF NOT EXISTS dev.gold")
print("Schema 'dev.gold' created")
```

```python
# show all schema in dev catelog
display(spark.sql("SHOW SCHEMAS IN dev"))
```
